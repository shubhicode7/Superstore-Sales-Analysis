# Business Analysis

```python
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)    #category wise sales
```

```python
df.groupby('Region')['Profit'].sum().sort_values(ascending=False)  #region wise profit
```

```python
df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)    #top 10 product
```

```python
df.groupby('Segment')['Sales'].sum()  #segment wise sales
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')          #monthly sales trend


monthly_sales = df.groupby('Month')['Sales'].sum()

print(monthly_sales)
```

# Visualization

```python
df.groupby('Category')['Sales'].sum().plot(            #category wise sales chart
    kind='bar',
    title='Category Wise Sales'
)

plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

```python
df.groupby('Region')['Profit'].sum().plot(                    #region wise profit chart
    kind='bar',
    title='Region Wise Profit'
)

plt.xlabel('Region')
plt.ylabel('Total Profit')
plt.show()
```

```python
monthly_sales.plot(                                         #monthly trend sales
    kind='line',
    marker='o',
    title='Monthly Sales Trend'
)

plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

# Business insights

```python
# 1. Category-wise sales and profit
category_analysis = df.groupby('Category').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).round(2)

display(category_analysis)

# 2. Region-wise performance
region_analysis = df.groupby('Region').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).round(2)

display(region_analysis)

# 3. Top 10 profitable products
top_products = df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).head(10)

display(top_products)

# 4. Loss-making products
loss_products = df.groupby('Product Name')['Profit'].sum().sort_values().head(10)

display(loss_products)
```

# Monthly sales trend

```python
monthly_sales = df.groupby('Month')['Sales'].sum()

monthly_sales.plot(
    kind='line',
    marker='o',
    figsize=(10,5),
    title='Monthly Sales Trend'
)

plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```
